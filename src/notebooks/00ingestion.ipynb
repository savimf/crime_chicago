## Police Stations

```python
client = Socrata('data.cityofchicago.org', None)
```

```python
results_stations = client.get(
    'z8bn-74gv',
    select="district, district_name, latitude, longitude"
)
print('Download complete!')

df_stations = pd.DataFrame.from_records(results_stations)
df_stations = f.rm_columns(df_stations)

print('POLICE STATIONS complete!')

df_stations.head()
```

```python
df_stations.to_csv(f'{brz_path}/df_stations.csv', index=False)
```

## Community Area

```python
results_comm = client.get(
    'igwz-8jzy',
    select='the_geom, area_numbe, community, shape_area, shape_len'
)
print('Download complete!')

df_comm = pd.DataFrame.from_records(results_comm)
df_comm = f.rm_columns(df_comm)

print('COMMUNITY AREAS complete!')

df_comm.head()
```

## Census

```python
results_census = client.get(
    'kn9c-c2s2',
    select="""
        ca, percent_of_housing_crowded,
        percent_households_below_poverty,
        percent_aged_16_unemployed,
        percent_aged_25_without_high_school_diploma,
        percent_aged_under_18_or_over_64,
        per_capita_income_,
        hardship_index
    """
)
print('Download complete!')

df_census = pd.DataFrame.from_records(results_census)
df_census = f.rm_columns(df_census)

print('CENSUS complete!')

df_census.head()
```

```python
df_census.to_csv(f'{brz_path}/df_census.csv', index=False)
```

## Districts

## Crimes

```python
results_crime = client.get(
    'ijzp-q8t2',
    select="""
            id, date, primary_type,
            arrest, domestic,
            district, community_area,
            latitude, longitude
        """,
    where="date >= '2022-01-01' and date < '2026-06-01'",
    limit=2_000_000
)
print('Download complete!')

df_crime = pd.DataFrame.from_records(results_crime)
df_crime = f.rm_columns(df_crime)

print('CRIMES complete!')

df_crime.head()
```

```python
df_crime.to_csv(f'{brz_path}/df_crime.csv', index=False)
```
